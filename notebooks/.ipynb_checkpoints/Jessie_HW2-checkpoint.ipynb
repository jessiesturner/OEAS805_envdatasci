```python
# Load in the CSV file, source:  https://data.norfolk.gov/
infile = '/Users/jsturner/OEAS805_envdatasci/data/International_Coastal_Cleanup.csv'

# read in the .csv file with the column separator explicitly defined
# handle comma thousands separators automatically (otherwise Cigarette Butts and Food Wrappers will not import properly!)
data = pd.read_csv(infile, sep = ',', thousands=",")
```

```python
# Considering Ocean View Beach Park had the most events, look at a time series there:
# Get just the year out of the date column:
# Convert 'Date' to datetime format, then extract the year
data['Year'] = pd.to_datetime(data['Date']).dt.year

# Create a variable for just Ocean View Beach Park
ov = data[data['Area'] == 'Ocean View Beach Park']
print(ov)
```

```python
sns.barplot(data = ov, x = 'Year', y = 'Cigarette Butts')
```

```python
# Compare multiple neighborhoods across one year, Total Pounds of Litter Collected
data['Year'].unique()
```

```python
# Find which year had the most neighborhoods
neighborhoods_per_year = data.groupby('Year')['Area'].nunique()
top_year = neighborhoods_per_year.idxmax()
max_count = neighborhoods_per_year.max()

print(f"The year with the most neighborhoods sampled was {top_year} with {max_count} unique locations.")
```

```python
oneyr = data[data['Year']==2016]
# print(oneyr)
```

```python
# Plot a histogram of the Food Wrappers category
plt.hist(data["Food Wrappers (candy, chips, etc.)"].dropna(), bins=12, color="steelblue", edgecolor="black")
plt.title("Histogram of Food Wrappers 2016 across 16 neighborhoods")
plt.xlabel("# of Food Wrappers")
plt.ylabel("Frequency")
plt.show()
```

```python
# Plot a bar graph of the Cigarette Butts across neighborhoods
fig, ax = plt.subplots(1, figsize = (20,8), dpi = 300) # Optional: specify figure size and resolution if you want
sns.barplot(data = oneyr, x = 'Area', y = 'Cigarette Butts')
ax.tick_params(axis='x', rotation=90)
plt.title("2016 Cleanup Events across 16 neighborhoods")
```

```python
# compare, for the five areas with the most events, the distribution of the different trash items across years in a boxplot
# Find the top 5 areas with the most cleanup events
top_5_areas = data["Area"].value_counts().nlargest(5).index.tolist()
```

```python
# Filter for only the top 5 areas
filtered_data = data[data["Area"].isin(top_5_areas)]
filtered_data['Year'].unique()
```

```python
# Plot the distribution across years grouped by area
plt.figure(figsize=(12, 6))
sns.boxplot(data=filtered_data,x="Area",y="Cigarette Butts")

# Customize plot aesthetics
plt.xlabel("Area", fontsize=12, labelpad=10)
plt.ylabel("Cigarette Butts Count", fontsize=12)
plt.xticks(rotation=15, ha="right")
plt.tight_layout()

plt.show()
```
